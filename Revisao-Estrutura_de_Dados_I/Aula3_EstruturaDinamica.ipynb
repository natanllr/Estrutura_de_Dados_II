## 5. Compartilhamento de referências

```python
a = [10, 20, 30]
b = a

print(id(a))
print(id(b))
```

```text
        ┌────────────────┐
        │ [10, 20, 30]   │
        └────────────────┘
             ▲       ▲
             │       │
             a       b
```

## 6. Referência versus cópia

### Referência

```python
a = [10, 20, 30]
b = a
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

### Cópia

```python
a = [10, 20, 30]
b = a.copy()
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

## 10. Nó (`Node`)

```python
class Node:

    def __init__(self, valor):
        self.valor = valor
        self.proximo = None
```

O atributo `proximo` é uma referência para outro nó.

## 11. Criando e conectando nós

```python
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.proximo = node2
node2.proximo = node3
```

Representação:

```text
10 ───→ 20 ───→ 30 ───→ None
```

## 12. Percorrendo os nós

```python
atual = node1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
```

Saída:

```text
10
20
30
```

## 13. Lista encadeada

```python
class ListaEncadeada:

    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = Node(valor)
        novo.proximo = self.inicio
        self.inicio = novo

    def exibir(self):
        atual = self.inicio

        while atual is not None:
            print(atual.valor)
            atual = atual.proximo
```

Uso:

```python
lista = ListaEncadeada()
lista.adicionar(10)
lista.adicionar(20)
lista.adicionar(30)
lista.exibir()
```

## 21. Funcionalidades obrigatórias

    class FilaAtendimento:
        def __init__(self):
            self.inicio = None
            self.fim = None
            self.quantidade = 0

        def adicionar(self, paciente):
            novo = Node(paciente)

            if self.esta_vazia():
                self.inicio = novo
                self.fim = novo
            else:
                self.fim.proximo = novo
                self.fim = novo

            self.quantidade += 1

        def atender(self):
            if self.esta_vazia():
                return None

            paciente = self.inicio.paciente
            self.inicio = self.inicio.proximo

            if self.inicio is None:
                self.fim = None

            self.quantidade -= 1

            return paciente

        def listar(self):
            atual = self.inicio

            while atual is not None:
                paciente = atual.paciente

                print(
                    paciente.nome,
                    "-",
                    paciente.idade,
                    "anos -",
                    paciente.prioridade
                )

                atual = atual.proximo

        def esta_vazia(self):
            return self.inicio is None

        def tamanho(self):
            return self.quantidade

## 27. Reflexão final

### O que é uma referência em Python?

Uma referência é uma variável que aponta para um objeto existente na memória. Quando duas variáveis fazem referência ao mesmo objeto, uma alteração nesse objeto pode ser percebida pelas duas variáveis.

### Qual a diferença entre `b = a` e `b = a.copy()`?

`b = a` faz `b` apontar para o mesmo objeto que `a`. Já `b = a.copy()` cria uma nova lista, independente da original.

### Por que uma estrutura encadeada precisa de referências?

Porque cada nó precisa guardar uma referência para o próximo nó. Dessa forma, os elementos podem ser conectados formando uma sequência dinâmica.

### Qual erro você encontrou durante a depuração?

Um erro foi não atualizar a variável `atual` durante o `while`. Isso fazia o programa permanecer no mesmo nó e entrar em um loop infinito.

A correção foi adicionar:

    atual = atual.proximo

### Como referências ajudam a compreender estruturas dinâmicas?

As referências permitem conectar objetos diferentes na memória. Em uma lista encadeada, cada `Node` utiliza uma referência para apontar para o próximo nó, permitindo criar, percorrer e modificar a estrutura dinamicamente.
